```python
def read_daily_log(sheet):
    records = []
    for row in range(6, sheet.max_row + 1):
        if sheet.cell(row, 1).value is None:
            continue
        record = {
            "date": sheet.cell(row, 1).value,
            "sleep": sheet.cell(row, 2).value,
            "fitness": sheet.cell(row, 3).value,
            "study": sheet.cell(row, 4).value,
            "coding": sheet.cell(row, 5).value,
            "class": sheet.cell(row, 6).value,
            "classes_attended": sheet.cell(row, 7).value,
            "other": sheet.cell(row, 8).value,
            "feeling": sheet.cell(row, 11).value,
            "satisfaction": sheet.cell(row, 12).value,
            "energy": sheet.cell(row, 13).value,
            "notes": sheet.cell(row, 14).value
        }
        records.append(record)
    return records

records = read_daily_log(sheet)
print("Number of records:", len(records))
```

CALCULATE TOTAL TRACKED RECORDS

```python
def calculate_total_tracked(record):

    total = (
        record["sleep"] + record["fitness"]+ record["study"]
        + record["coding"]+ record["class"]+ record["other"]
    )

    return total
print(calculate_total_tracked(records[0]) ,"min")
```

```python
def calculate_free_time(record):

    total_tracked = calculate_total_tracked(record)

    free_time = 1440 - total_tracked

    return free_time

print(calculate_free_time(records[0]),"mins")
```

```python
def average_value(records, field):

    total = 0

    for record in records:
        total = total + record[field]

    return total / len(records)
```

7. DCI — Data Continuity Index


```python
from datetime import date, datetime

def calculate_dci(records):
    start_date = date(2026, 8, 13)
    end_date = date(2026, 9, 25)

    expected_days = (end_date - start_date).days + 1

    recorded_days = set()

    for record in records:
        record_date = record["date"]


        if isinstance(record_date, datetime):
            record_date = record_date.date()

        if start_date <= record_date <= end_date:
            recorded_days.add(record_date)

    valid_recorded_days = len(recorded_days)

    dci = (valid_recorded_days / expected_days) * 100

    return dci


dci = calculate_dci(records)

print("Data Continuity Index (DCI):", round(dci, 2), "%")
```

8. EI — Experience Index

```python
def calculate_ei(records):

    feeling_scores = {
        "Excellent": 5,
        "Good": 4,
        "Neutral": 3,
        "Low": 2,
        "Stressed": 1
    }

    satisfaction_scores = {
        "Very Satisfied": 5,
        "Satisfied": 4,
        "Neutral": 3,
        "Unsatisfied": 2,
        "Very Unsatisfied": 1
    }

    energy_scores = {
        "High": 3,
        "Medium": 2,
        "Low": 1
    }

    total_score = 0

    for record in records:
        feeling = feeling_scores[record["feeling"]]
        satisfaction = satisfaction_scores[record["satisfaction"]]
        energy = energy_scores[record["energy"]]

        total_score = total_score + feeling + satisfaction + energy

    ei = total_score / (3 * len(records))

    return ei


ei = calculate_ei(records)

print("Experience Index (EI):", round(ei, 2), "/ 5")
```

Personal Activity Index (PAI)

```python
def calculate_pai(tpi, aai, phai, sri, tui, ei, dci):
    pai = (
        0.15 * tpi
        + 0.20 * aai
        + 0.15 * phai
        + 0.20 * sri
        + 0.15 * tui
        + 0.10 * ei
        + 0.05 * dci
    )

    return pai


pai = calculate_pai(tpi, aai, phai, sri, tui, ei, dci)

print("Personal Activity Index (PAI):", round(pai, 2))
```

correlation function

```python
import math

def corr(x, y):
    n = len(x)

    mx = sum(x) / n
    my = sum(y) / n

    a = 0
    b = 0
    c = 0

    for i in range(n):
        a += (x[i] - mx) * (y[i] - my)
        b += (x[i] - mx) ** 2
        c += (y[i] - my) ** 2

    return a / math.sqrt(b * c)
```
