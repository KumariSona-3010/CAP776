```python
from openpyxl import load_workbook

# Ensure workbook and sheet are loaded
try:
    workbook
except NameError:
    file_path = '/content/drive/MyDrive/Sonali_12610356.xlsx'
    workbook = load_workbook(file_path)

sheet = workbook["Daily Log"]

print("Workbook and Daily Log sheet loaded successfully")
print("Rows:", sheet.max_row)
print("Columns:", sheet.max_column)




```

```python
for column in range(1, 15):
    print(column, sheet.cell(5, column).value)
```

```python
def read_daily_log(sheet):
    records = []
    for row in range(6, sheet.max_row + 1):
        if sheet.cell(row, 1).value is None:
            continue
        record = {
            "date": sheet.cell(row, 1).value,
            "sleep": sheet.cell(row, 2).value,
            "fitness": sheet.cell(row, 3).value,
            "study": sheet.cell(row, 4).value,
            "coding": sheet.cell(row, 5).value,
            "class": sheet.cell(row, 6).value,
            "classes_attended": sheet.cell(row, 7).value,
            "other": sheet.cell(row, 8).value,
            "feeling": sheet.cell(row, 11).value,
            "satisfaction": sheet.cell(row, 12).value,
            "energy": sheet.cell(row, 13).value,
            "notes": sheet.cell(row, 14).value
        }
        records.append(record)
    return records

records = read_daily_log(sheet)
print("Number of records:", len(records))
```

CALCULATE TOTAL TRACKED RECORDS

```python
def calculate_total_tracked(record):

    total = (
        record["sleep"] + record["fitness"]+ record["study"]
        + record["coding"]+ record["class"]+ record["other"]
    )

    return total
print(calculate_total_tracked(records[0]) ,"min")
```

CALCULATE UNACCOUNTED MINUTES
****

```python
def calculate_free_time(record):

    total_tracked = calculate_total_tracked(record)

    free_time = 1440 - total_tracked

    return free_time

print(calculate_free_time(records[0]),"mins")
```

Add calculated values to every record

```python
for record in records:

    record["total_tracked"] = calculate_total_tracked(record)
    record["free_time"] = calculate_free_time(record)
print(records[0],"mins")
```

```python
def average_value(records, field):

    total = 0

    for record in records:
        total = total + record[field]

    return total / len(records)
```

. TPI — Tech Productivity Index

```python
def calculate_tpi(records):
    total_coding = 0

    for record in records:
        total_coding = total_coding + record["coding"]

    tpi = total_coding / len(records)

    return tpi


tpi = calculate_tpi(records)

print("Tech Productivity Index (TPI):", round(tpi, 2), "min/day")
```

AAI — Academic Activity Index


```python
def calculate_aai(records):
    total_academic = 0

    for record in records:
        total_academic = total_academic + record["study"] + record["class"]

    aai = total_academic / len(records)

    return aai


aai = calculate_aai(records)

print("Academic Activity Index (AAI):", round(aai, 2), "min/day")
```

3. PhAI — Physical Activity Index

```python
def calculate_phai(records):
    total_fitness = 0

    for record in records:
        total_fitness = total_fitness + record["fitness"]

    phai = total_fitness / len(records)

    return phai


phai = calculate_phai(records)

print("Physical Activity Index (PhAI):", round(phai, 2), "min/day")
```

4. SRI — Sleep & Recovery Index

```python
def calculate_sri(records):
    total_sleep = 0

    for record in records:
        total_sleep = total_sleep + record["sleep"]

    sri = total_sleep / len(records)

    return sri


sri = calculate_sri(records)

print("Sleep & Recovery Index (SRI):", round(sri, 2), "min/day")
```

5. ABI — Activity Balance Index

```python
def calculate_abi(records):
    total_free_time = 0

    for record in records:
        total_free_time = total_free_time + record["free_time"]

    abi = total_free_time / len(records)

    return abi


abi = calculate_abi(records)

print("Activity Balance Index (ABI):", round(abi, 2), "min/day")
```

6. TUI — Time Utilization Index

```python
def calculate_tui(records):
    total_tracked_time = 0

    for record in records:
        total_tracked_time = total_tracked_time + record["total_tracked"]

    tui = total_tracked_time / len(records)

    return tui


tui = calculate_tui(records)

print("Time Utilization Index (TUI):", round(tui, 2), "min/day")
```
