```python
from openpyxl import load_workbook

# Ensure workbook and sheet are loaded
try:
    workbook
except NameError:
    file_path = '/content/drive/MyDrive/Sonali_12610356.xlsx'
    workbook = load_workbook(file_path)

sheet = workbook["Daily Log"]

print("Workbook and Daily Log sheet loaded successfully")
print("Rows:", sheet.max_row)
print("Columns:", sheet.max_column)




```

```python
def read_daily_log(sheet):
    records = []
    for row in range(6, sheet.max_row + 1):
        if sheet.cell(row, 1).value is None:
            continue
        record = {
            "date": sheet.cell(row, 1).value,
            "sleep": sheet.cell(row, 2).value,
            "fitness": sheet.cell(row, 3).value,
            "study": sheet.cell(row, 4).value,
            "coding": sheet.cell(row, 5).value,
            "class": sheet.cell(row, 6).value,
            "classes_attended": sheet.cell(row, 7).value,
            "other": sheet.cell(row, 8).value,
            "feeling": sheet.cell(row, 11).value,
            "satisfaction": sheet.cell(row, 12).value,
            "energy": sheet.cell(row, 13).value,
            "notes": sheet.cell(row, 14).value
        }
        records.append(record)
    return records

records = read_daily_log(sheet)
print("Number of records:", len(records))
```

CALCULATE TOTAL TRACKED RECORDS

```python
def calculate_total_tracked(record):

    total = (
        record["sleep"] + record["fitness"]+ record["study"]
        + record["coding"]+ record["class"]+ record["other"]
    )

    return total
print(calculate_total_tracked(records[0]) ,"min")
```
